```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create time values
t = np.arange(-5, 5, 0.01)

# 1. Unit Step Signal
unit_step = np.where(t >= 0, 1, 0)

# 2. Unit Impulse Signal (discrete approximation)
impulse = np.where(np.isclose(t, 0), 1, 0)

# 3. Ramp Signal
ramp = np.maximum(t, 0)

# 4. Sine Signal
sine = np.sin(2 * np.pi * t)

# 5. Cosine Signal
cosine = np.cos(2 * np.pi * t)

# Create graphs
plt.figure(figsize=(12, 10))

plt.subplot(5, 1, 1)
plt.plot(t, unit_step)
plt.title("Unit Step Signal")
plt.grid()

plt.subplot(5, 1, 2)
plt.stem(t[::10], impulse[::10])
plt.title("Unit Impulse Signal")
plt.grid()

plt.subplot(5, 1, 3)
plt.plot(t, ramp)
plt.title("Ramp Signal")
plt.grid()

plt.subplot(5, 1, 4)
plt.plot(t, sine)
plt.title("Sine Wave")
plt.grid()

plt.subplot(5, 1, 5)
plt.plot(t, cosine)
plt.title("Cosine Wave")
plt.grid()

plt.tight_layout()
plt.show()
```
